## 🌡️ 8. Temperature Unit Converter

### Goal:
    Build a small converter for a weather app.

### Steps:

    Show a menu: 1 = Celsius to Fahrenheit, 2 = Fahrenheit to Celsius.
    
    Ask for the choice and then the temperature.

    Any other choice → "Invalid choice".

### Example:

## 🚕 11. Taxi Fare Calculator

### Goal:
    Calculate the fare of a ride-booking app.

### Steps:

    Ask for the ride type (economy or luxury).
    
    Ask for the distance in km.
    
    Economy: Rs 200 for up to 5 km, then Rs 30 per extra km.
    
    Luxury: Rs 400 for up to 5 km, then Rs 60 per extra km.
    
    Print the final fare.

### Example:

    Enter ride type: economy
    Enter distance: 8
    Fare = Rs 290

```python
ride = input("Enter ride type: ").lower()
distance = float(input("Enter distance: "))

if ride == "economy":
    if distance <= 5:
        fare = 200
    else:
        fare = 200 + (distance - 5) * 30
    print("Fare = Rs", fare)

elif ride == "luxury":
    if distance <= 5:
        fare = 400
    else:
        fare = 400 + (distance - 5) * 60
    print("Fare = Rs", fare)

else:
    print("Invalid ride type")
```
